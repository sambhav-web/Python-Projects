```python
from IPython.display import clear_output
decision='yes'
print('Welcome to the Secret Auction!')
bidders={}
winner=""
max_bid=0
while decision=='yes':
  user=input("Name of the bidder: ")
  while True:
    try:
      bet=float(input('Enter the bid : $'))
      if bet<0:
        raise ValueError("Invalid Bet.")
      break
    except ValueError as exp:
      print(exp)
  bidders[user]=bet
  while True:
    try:
      decision=input("Are there any other bidder? (yes/no) :").lower()
      if decision not in ['yes','no']:
        raise ValueError("Invalid Decision.")
      break
    except ValueError as exp:
      print(exp)
  clear_output(wait=True)
  if decision=='no':
    for i,j in bidders.items():
      if j>max_bid:
        max_bid=j
        winner=i
    print(f'Participants are {bidders}')
    print(f"The winner of Auction is {winner} with maximum bid amount of {max_bid}")
    break
```
