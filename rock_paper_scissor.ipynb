```python
import random
print("Welcome to the Rock,Paper and Scissors Game!")
def choice():
  return random.randint(0,2)
def label(x):
  if x==0:
    return "Rock"
  elif x==1:
    return "Paper"
  else:
    return "Scissors"
def game(user,com,u,c):
  if (user,com) in [("Rock","Scissors"),("Paper","Rock"),("Scissors","Paper")]:
    return "😆,You wins!",u+1,c
  elif (com,user) in [("Rock","Scissors"),("Paper","Rock"),("Scissors","Paper")]:
    return "😞,You Lose!",u,c+1
  else:
    return "Draw",u,c
Decision="y"
user_score=0
com_score=0
while Decision=="y":
  com=choice()
  com=label(com)
  while True:
    try:
      user=int(input("Enter :\n 0 for Rock\n 1 for Paper\n and 2 for Scissors: "))
      if user not in range(3):
        raise ValueError("Invalid Choice")
      break
    except ValueError as exp:
      print(exp)
  user=label(user)
  print(f"You chose {user} and Computer chose {com}")
  result,user_score,com_score=game(user,com,user_score,com_score)
  print(result)
  print(f"Your Score is {user_score} and Computer Score is {com_score}")
  while True:
    try:
      Decision=input("Enter Y to play again and N to pass: ").lower()
      if Decision not in ["y","n"]:
        raise ValueError("Invalid Choice")
      break
    except ValueError as exp:
      print(exp)
```
