```python
from random import randint
def select_level(lvl):
  if lvl=="easy":
    return 10
  elif lvl=="medium":
    return 7
  elif lvl=="hard":
    return 5
  elif lvl=="insane":
    return 3
def check_guess(user,com,attempts,score):
  if user>com:
    return "Oops🥺,Too High!",attempts-1,score+1,True
  elif user<com:
    return "Oops🥺,Too Low!",attempts-1,score+1,True
  else:
    return "Congratulations😆, You guessed it right!",attempts-1,score+1,False
print("Welcome to the Number Guessing Game!")
com=randint(1,100)
while True:
  try:
    level=input("Choose Difficulty Level-\neasy\nmedium\nhard\ninsane: ")
    if level not in ['easy','hard','medium','insane']:
      raise ValueError("Invalid Level Choosen.")
    break
  except ValueError as e:
    print(e)
attempts=select_level(level)
score=0
is_on=True
print(f"---You have {attempts} attempts to guess the number correctly---")
while is_on:
  while True:
    try:
      user=int(input("Guess a number between 1 and 100: "))
      if user not in range(1,101):
        raise ValueError("Invalid Value.Plz choose within the specified range")
      break
    except ValueError as e:
      print(e)
  result,attempts,score,is_on=check_guess(user,com,attempts,score)
  print(result)
  if attempts==0:
    print(f"You have run out of attempts.The correct number was {com}")
    break
  else:
    print(f"You have {attempts} attempts left")
  if is_on==False:
    print(f"Your Score: {score}")
    break
```
